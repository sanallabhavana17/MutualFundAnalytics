```python
import pandas as pd

df = pd.read_csv(r"../Data/Raw/drive-download-20260724T173806Z-1-001/08_investor_transactions.csv")

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(data=df, x="transaction_type")

plt.title("Transaction Types")
plt.show()
```

```python
sns.countplot(data=df, x="payment_mode")

plt.title("Payment Modes")
plt.xticks(rotation=30)

plt.show()
```

```python
sns.countplot(data=df, x="kyc_status")

plt.title("KYC Status")

plt.show()
```

```python
top_states = df["state"].value_counts().head(10)

top_states.plot(kind="bar")

plt.title("Top 10 States")

plt.show()
```

```python
sns.histplot(df["amount_inr"], bins=30)

plt.title("Investment Amount Distribution")

plt.show()
```

```python
sns.countplot(data=df, x="age_group")

plt.title("Age Group Distribution")
plt.xlabel("Age Group")
plt.ylabel("Number of Investors")
plt.show()
```

```python
sns.countplot(data=df, x="gender")

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Investors")
plt.show()
```

```python
sns.countplot(data=df, x="city_tier")

plt.title("City Tier Distribution")
plt.xlabel("City Tier")
plt.ylabel("Number of Investors")
plt.show()
```

```python
sns.boxplot(data=df, x="transaction_type", y="amount_inr")

plt.title("Transaction Amount by Type")
plt.show()
```

```python
sns.histplot(df["annual_income_lakh"], bins=20)

plt.title("Annual Income Distribution")
plt.xlabel("Income (Lakhs)")
plt.show()
```

```python
avg = df.groupby("payment_mode")["amount_inr"].mean()

avg.plot(kind="bar")

plt.title("Average Investment by Payment Mode")
plt.ylabel("Average Amount")
plt.show()
```

```python
avg = df.groupby("age_group")["amount_inr"].mean()

avg.plot(kind="bar")

plt.title("Average Investment by Age Group")
plt.ylabel("Average Amount")
plt.show()
```

```python
top = df["city"].value_counts().head(10)

top.plot(kind="bar")

plt.title("Top 10 Cities")
plt.show()
```

```python
df["payment_mode"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Payment Mode Distribution")
plt.ylabel("")
plt.show()
```

```python
df["kyc_status"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("KYC Status Distribution")
plt.ylabel("")
plt.show()
```
