```python
import pandas as pd
import matplotlib.pyplot as plt
import glob
import os

nav_files = glob.glob("../Data/Raw/*_nav.csv")

nav_data = []

for file in nav_files:
    df_nav = pd.read_csv(file)
    
    # Add fund code from filename
    fund_code = os.path.basename(file).replace("_nav.csv", "")
    df_nav["fund_code"] = fund_code
    
    nav_data.append(df_nav)

nav_df = pd.concat(nav_data, ignore_index=True)

print("Total NAV records:", len(nav_df))
print("Columns:", nav_df.columns.tolist())
print(nav_df.head())
```

```python
# Convert date column to datetime
nav_df["date"] = pd.to_datetime(
    nav_df["date"],
    dayfirst=True,
    errors="coerce"
)

# Convert NAV to numeric
nav_df["nav"] = pd.to_numeric(
    nav_df["nav"],
    errors="coerce"
)

# Remove invalid rows
nav_df = nav_df.dropna(subset=["date", "nav"])

# Sort by fund and date
nav_df = nav_df.sort_values(["fund_code", "date"])

print("Date range:")
print(nav_df["date"].min(), "to", nav_df["date"].max())

print("\nTotal records after cleaning:", len(nav_df))

print("\nRecords by fund:")
print(nav_df["fund_code"].value_counts().sort_index())
```

```python
# Filter NAV data for 2022–2026
nav_2022_2026 = nav_df[
    (nav_df["date"] >= "2022-01-01") &
    (nav_df["date"] <= "2026-12-31")
].copy()

# Create NAV trend chart
plt.figure(figsize=(14, 7))

for fund in nav_2022_2026["fund_code"].unique():
    fund_data = nav_2022_2026[
        nav_2022_2026["fund_code"] == fund
    ].sort_values("date")

    plt.plot(
        fund_data["date"],
        fund_data["nav"],
        label=fund,
        linewidth=1.5
    )

# Highlight 2023 bull-run period
plt.axvspan(
    pd.Timestamp("2023-01-01"),
    pd.Timestamp("2023-12-31"),
    alpha=0.10,
    label="2023 Bull Run"
)

# Highlight 2024 correction period
plt.axvspan(
    pd.Timestamp("2024-01-01"),
    pd.Timestamp("2024-12-31"),
    alpha=0.10,
    label="2024 Correction"
)

plt.title("NAV Trend Analysis — 2022 to 2026")
plt.xlabel("Date")
plt.ylabel("NAV")
plt.legend(title="Fund Code")
plt.grid(True, alpha=0.3)
plt.tight_layout()

# Save PNG
plt.savefig(
    "../Reports/EDA_Charts/nav_trend_2022_2026.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert date to datetime
aum_df["date"] = pd.to_datetime(aum_df["date"])

# Extract year
aum_df["year"] = aum_df["date"].dt.year

# Keep 2022–2025
aum_yearly = aum_df[aum_df["year"].between(2022, 2025)]

# Take the latest AUM value available for each fund house in each year
aum_year_end = (
    aum_yearly
    .sort_values("date")
    .groupby(["year", "fund_house"])
    .tail(1)
)

print(aum_year_end[["year", "fund_house", "aum_lakh_crore"]].head(20))
```

```python
plt.figure(figsize=(14, 7))

sns.barplot(
    data=aum_year_end,
    x="year",
    y="aum_lakh_crore",
    hue="fund_house"
)

plt.title("AUM Growth by Fund House (2022–2025)")
plt.xlabel("Year")
plt.ylabel("AUM (₹ Lakh Crore)")
plt.legend(title="Fund House", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()

plt.savefig(
    "../Reports/EDA_Charts/aum_growth_by_fund_house.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import pandas as pd
import plotly.express as px

# Load SIP inflow dataset
sip_df = pd.read_csv(
    "../Data/Raw/04_monthly_sip_inflows.csv"
)

# Convert month to datetime
sip_df["month"] = pd.to_datetime(sip_df["month"])

# Create SIP inflow trend chart
fig = px.line(
    sip_df,
    x="month",
    y="sip_inflow_crore",
    title="Monthly SIP Inflow Trend (2022–2025)",
    labels={
        "month": "Month",
        "sip_inflow_crore": "SIP Inflow (₹ Crore)"
    }
)

# Highlight maximum SIP inflow
max_row = sip_df.loc[sip_df["sip_inflow_crore"].idxmax()]

fig.add_annotation(
    x=max_row["month"],
    y=max_row["sip_inflow_crore"],
    text=f"₹{max_row['sip_inflow_crore']:,.0f} Cr all-time high",
    showarrow=True,
    arrowhead=2
)

# Display chart
fig.show()

# Save chart as PNG
fig.write_image(
    "../Reports/EDA_Charts/sip_inflow_trend.png",
    width=1200,
    height=600
)

print("SIP chart saved successfully!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Load SIP dataset
sip_df = pd.read_csv("../Data/Raw/04_monthly_sip_inflows.csv")

# Convert month to date
sip_df["month"] = pd.to_datetime(sip_df["month"])

# Create chart
plt.figure(figsize=(12, 6))

plt.plot(
    sip_df["month"],
    sip_df["sip_inflow_crore"],
    linewidth=2
)

plt.title("Monthly SIP Inflow Trend")
plt.xlabel("Month")
plt.ylabel("SIP Inflow (Crore)")

plt.xticks(rotation=45)
plt.tight_layout()

# Save chart
output_path = "../Reports/EDA_Charts/sip_inflow_trend.png"
plt.savefig(output_path, dpi=300, bbox_inches="tight")

plt.show()

print("SIP chart saved successfully!")
print("Location:", os.path.abspath(output_path))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Load Category Inflows dataset
category_df = pd.read_csv("../Data/Raw/05_category_inflows.csv")

# Total net inflow by category
category_total = (
    category_df.groupby("category")["net_inflow_crore"]
    .sum()
    .sort_values(ascending=False)
)

print(category_total)

# Create chart
plt.figure(figsize=(12, 6))

category_total.plot(kind="bar")

plt.title("Total Net Inflow by Mutual Fund Category")
plt.xlabel("Category")
plt.ylabel("Net Inflow (Crore)")
plt.xticks(rotation=45)
plt.tight_layout()

# Save chart
output_path = "../Reports/EDA_Charts/category_inflows.png"

plt.savefig(output_path, dpi=300, bbox_inches="tight")

plt.show()

print("Category inflow chart saved successfully!")
print("Location:", os.path.abspath(output_path))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Load scheme performance dataset
scheme_df = pd.read_csv("../Data/Raw/07_scheme_performance.csv")

# Calculate average 1-year return by category
category_returns = (
    scheme_df.groupby("category")["return_1yr_pct"]
    .mean()
    .sort_values(ascending=False)
)

print(category_returns)

# Create chart
plt.figure(figsize=(12, 6))

category_returns.plot(kind="bar")

plt.title("Average 1-Year Return by Mutual Fund Category")
plt.xlabel("Fund Category")
plt.ylabel("Average 1-Year Return (%)")
plt.xticks(rotation=45)
plt.tight_layout()

# Save chart
output_path = "../Reports/EDA_Charts/scheme_category_returns.png"

plt.savefig(output_path, dpi=300, bbox_inches="tight")

plt.show()

print("Scheme performance chart saved successfully!")
print("Location:", os.path.abspath(output_path))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Load portfolio holdings
portfolio_df = pd.read_csv("../Data/Raw/09_portfolio_holdings.csv")

# Calculate average sector allocation
sector_weight = (
    portfolio_df.groupby("sector")["weight_pct"]
    .mean()
    .sort_values(ascending=False)
)

print(sector_weight)

# Create chart
plt.figure(figsize=(12, 6))
sector_weight.plot(kind="bar")

plt.title("Average Portfolio Allocation by Sector")
plt.xlabel("Sector")
plt.ylabel("Average Weight (%)")
plt.xticks(rotation=45)
plt.tight_layout()

# Save
output_path = "../Reports/EDA_Charts/portfolio_sector_allocation.png"

plt.savefig(output_path, dpi=300, bbox_inches="tight")
plt.close()

print("Portfolio sector chart saved successfully!")
print("Location:", os.path.abspath(output_path))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Load benchmark dataset
benchmark_df = pd.read_csv("../Data/Raw/10_benchmark_indices.csv")

# Convert date to datetime
benchmark_df["date"] = pd.to_datetime(benchmark_df["date"])

# Create chart
plt.figure(figsize=(12, 6))

# Plot each benchmark index
for index_name in benchmark_df["index_name"].unique():
    index_data = benchmark_df[benchmark_df["index_name"] == index_name]

    plt.plot(
        index_data["date"],
        index_data["close_value"],
        linewidth=1.5,
        label=index_name
    )

plt.title("Benchmark Index Performance")
plt.xlabel("Date")
plt.ylabel("Index Closing Value")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()

# Save chart
output_path = "../Reports/EDA_Charts/benchmark_index_trend.png"

plt.savefig(output_path, dpi=300, bbox_inches="tight")
plt.close()

print("Benchmark index chart saved successfully!")
print("Location:", os.path.abspath(output_path))
```
