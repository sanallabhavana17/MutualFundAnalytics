# Key EDA Findings

### Finding 1
SIP transactions are the most common transaction type among investors.

### Finding 2
Most investors have completed KYC verification.

### Finding 3
UPI, Net Banking and Mandate are the most frequently used payment methods.

### Finding 4
Most investment transactions come from Tier-1 (T30) cities.

### Finding 5
The age group 26–35 contributes the highest number of investors.

### Finding 6
Male investors are slightly higher than female investors.

### Finding 7
Some states contribute significantly more investments than others.

### Finding 8
Lumpsum investments have a higher average investment amount than SIP.

### Finding 9
Verified investors contribute most of the investment amount.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../Data/Raw/drive-download-20260724T173806Z-1-001/04_monthly_sip_inflows.csv")

df["month"] = pd.to_datetime(df["month"])

plt.figure(figsize=(12,6))

plt.plot(df["month"], df["sip_inflow_crore"],
         marker="o",
         linewidth=2)

# Highlight maximum SIP inflow
max_row = df.loc[df["sip_inflow_crore"].idxmax()]

plt.scatter(max_row["month"],
            max_row["sip_inflow_crore"],
            color="red",
            s=120)

plt.text(max_row["month"],
         max_row["sip_inflow_crore"]+300,
         f'Highest: {max_row["sip_inflow_crore"]} Cr',
         color="red")

plt.title("Monthly SIP Inflow Trend (2022-2025)")
plt.xlabel("Month")
plt.ylabel("SIP Inflow (Crore)")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Read the dataset
df = pd.read_csv("../Data/Raw/drive-download-20260724T173806Z-1-001/05_category_inflows.csv")

# Convert month to datetime
df["month"] = pd.to_datetime(df["month"])

# Create pivot table
heatmap_data = df.pivot(
    index="category",
    columns="month",
    values="net_inflow_crore"
)

# Plot heatmap
plt.figure(figsize=(14,8))

sns.heatmap(
    heatmap_data,
    cmap="YlGnBu",
    annot=True,
    fmt=".0f"
)

plt.title("Category-wise Net Inflow Heatmap")
plt.xlabel("Month")
plt.ylabel("Fund Category")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Read the dataset
df = pd.read_csv("../Data/Raw/drive-download-20260724T173806Z-1-001/06_industry_folio_count.csv")

# Select only numeric columns
numeric_df = df.select_dtypes(include=["number"])

# Create correlation matrix
corr = numeric_df.corr()

# Plot heatmap
plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix of Industry Folio Data")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the dataset
df = pd.read_csv("../Data/Raw/drive-download-20260724T173806Z-1-001/09_portfolio_holdings.csv")

# Group by sector
sector_data = df.groupby("sector")["weight_pct"].sum()

# Create Donut Chart
plt.figure(figsize=(8,8))

plt.pie(
    sector_data,
    labels=sector_data.index,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width":0.4}
)

plt.title("Sector Allocation of Portfolio")

plt.show()
```
