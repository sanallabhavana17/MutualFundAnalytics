# REST API & JSON Data Extraction

## Objective

The objective of this assignment is to call a public REST API, inspect the JSON response, convert the JSON data into a structured DataFrame, and export the data into a CSV file for analysis.

## Tools Used

- Python
- Requests
- Pandas
- Jupyter Notebook
- REST API
- JSON

```python
import requests
import pandas as pd

print("Libraries imported successfully!")
```

```python
# Public Mutual Fund API

url = "https://api.mfapi.in/mf/119551"

response = requests.get(url)

print("Status Code:", response.status_code)
```

```python
# Convert the API response into JSON

data = response.json()

print(type(data))
print(data.keys())
```

```python
# Extract the NAV data from the JSON response

nav_data = data["data"]

print("Number of records:", len(nav_data))
print("\nFirst 5 records:")
print(nav_data[:5])
```

```python
# Convert JSON data into a Pandas DataFrame

df = pd.DataFrame(nav_data)

print("DataFrame created successfully!")
print("\nShape:", df.shape)

df.head()
```

```python
# Save the API data as a CSV file

df.to_csv("api_data.csv", index=False)

print("CSV file created successfully!")
```

```python
# Verify the exported CSV file

csv_df = pd.read_csv("api_data.csv")

print("CSV loaded successfully!")
print("Rows:", len(csv_df))
print("Columns:", list(csv_df.columns))

csv_df.head()
```

## Conclusion

The API returned mutual fund NAV information in JSON format. The JSON response 
was inspected and the required data was extracted using Python. The extracted
